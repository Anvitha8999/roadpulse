```python
   from ultralytics import YOLO

   model = YOLO("yolo26n-seg.pt")
   model.train(data="crack-seg.yaml", epochs=20, imgsz=640, batch=16, patience=5)
   print("BEST WEIGHTS:", model.trainer.best)
```

```python
best = YOLO(model.trainer.best)
m = best.val(data="crack-seg.yaml", split="test")
print("TEST mask mAP50:", round(m.seg.map50, 3), "| mask mAP50-95:", round(m.seg.map, 3))

from google.colab import files
files.download(str(model.trainer.best))
```
